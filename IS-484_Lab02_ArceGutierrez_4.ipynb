<h2 style="color:#1e4a76; margin-top:24px;">Descripción del caso</h2>

<div style="background:#f0f7ff; border-left:6px solid #2a7de1; padding:16px 20px; border-radius:0 12px 12px 0; box-shadow: 0 2px 8px rgba(42,125,225,0.08); margin-bottom:20px;">
  <p style="margin:0; color:#1a3b5c;">
    Un repartidor debe ir desde el <b>Almacén</b> hasta la <b>Zona Norte</b>, pudiendo pasar
    por distintos puntos intermedios de la ciudad. Se aplica <b>DFS</b> para encontrar
    <b>un camino</b> (no necesariamente el más corto) explorando en profundidad antes de retroceder.
  </p>
</div>
