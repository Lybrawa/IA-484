<h2 style="color:#1e4a76; margin-top:24px;">Parte C: Implementación de DFS (Búsqueda en Profundidad)</h2>

<div style="background:#f0f7ff; border-left:6px solid #2a7de1; padding:16px 20px; border-radius:0 12px 12px 0; box-shadow: 0 2px 8px rgba(42,125,225,0.08); margin-bottom:20px;">
  <p style="margin:0; color:#1a3b5c;">
    DFS explora el grafo <b>en profundidad</b>: avanza por una rama hasta el final y luego
    <b>retrocede</b> (backtracking) para explorar las alternativas. Se implementa con una
    <b>pila LIFO</b> (<code style="background:#eef4fc; color:#1a5ca8; padding:2px 6px; border-radius:4px;">lista.pop()</code>)

  </p>
</div>

<h2 style="color:#1e4a76; margin-top:28px;">Pseudocódigo</h2>

<div style="background:#0f2233; color:#d6e6ff; padding:18px 22px; border-radius:12px; font-family:'JetBrains Mono','Consolas',monospace; font-size:0.92em; line-height:1.7; box-shadow: 0 4px 12px rgba(15,34,51,0.25);">
  <span style="color:#7fb4ff;">DFS</span>(grafo, inicio, objetivo):<br>
  &nbsp;&nbsp;pila ← [[inicio]]<br>
  &nbsp;&nbsp;visitados ← ∅<br>
  &nbsp;&nbsp;<span style="color:#7fb4ff;">mientras</span> pila ≠ ∅:<br>
  &nbsp;&nbsp;&nbsp;&nbsp;camino ← pila.<span style="color:#9de0bd;">pop()</span><br>
  &nbsp;&nbsp;&nbsp;&nbsp;nodo_actual ← camino[-1]<br>
  &nbsp;&nbsp;&nbsp;&nbsp;<span style="color:#7fb4ff;">si</span> nodo_actual = objetivo: <span style="color:#7fb4ff;">retornar</span> camino<br>
  &nbsp;&nbsp;&nbsp;&nbsp;<span style="color:#7fb4ff;">si</span> nodo_actual ∉ visitados:<br>
  &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;visitados.<span style="color:#9de0bd;">add</span>(nodo_actual)<br>
  &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;<span style="color:#7fb4ff;">para</span> vecino en grafo[nodo_actual]:<br>
  &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;pila.<span style="color:#9de0bd;">append</span>(camino + [vecino])<br>
  &nbsp;&nbsp;<span style="color:#7fb4ff;">retornar</span> None
</div>

</div>

<div style="font-family: 'Segoe UI', Roboto, Arial, sans-serif; background: linear-gradient(135deg,#e8f0fe 0%,#f3f7fb 100%); padding: 20px 24px; border-radius: 16px; color:#1a2b3c;">

<h2 style="color:#1e4a76; margin:0; font-weight:600;">
Código — Búsqueda en profundidad (DFS)
</h2>

</div>

```python
def dfs(grafo, inicio, objetivo):
    """
    Busca un camino entre 'inicio' y 'objetivo' usando DFS.
    No garantiza el camino más corto, pero explora en profundidad.
    """
    pila = [[inicio]]         
    visitados = set()

    while pila:
        camino = pila.pop()   
        nodo_actual = camino[-1]

        if nodo_actual == objetivo:
            return camino

        if nodo_actual not in visitados:
            visitados.add(nodo_actual)

            for vecino in grafo[nodo_actual]:
                nuevo_camino = camino + [vecino]
                pila.append(nuevo_camino)

    return None  

resultado = dfs(mapa, "Almacen", "ZonaNorte")
print("Camino encontrado (DFS):", resultado)
```

  <p style="margin:0; color:#1a3b5c;">

    como estrategia para escribir código más limpio, legible y rápido.

<h2 style="color:#1e4a76; margin-top:28px;">4. Aprendizajes clave</h2>

<div style="background:#f0f7ff; border-left:6px solid #2a7de1; padding:18px 22px; border-radius:0 12px 12px 0; box-shadow: 0 2px 8px rgba(42,125,225,0.08); margin-bottom:14px;">
  <p style="margin:0; color:#1a3b5c;">
    <b>① Los datos estructurados son el punto de partida.</b> Sin una correcta
    representación como vectores y matrices, cualquier análisis posterior (similitud,
    clustering, búsqueda) pierde sustento. NumPy ofrece las primitivas para esa
    representación de forma natural.
  </p>
</div>

<div style="background:#f0f7ff; border-left:6px solid #2a7de1; padding:18px 22px; border-radius:0 12px 12px 0; box-shadow: 0 2px 8px rgba(42,125,225,0.08); margin-bottom:14px;">
  <p style="margin:0; color:#1a3b5c;">
    <b>② Las métricas vectoriales cuantifican relaciones.</b> El producto escalar y la
    distancia euclidiana permiten comparar observaciones numéricamente, transformando
    datos crudos en información útil para tareas de decisión y clasificación.
  </p>
</div>

<div style="background:#f0f7ff; border-left:6px solid #2a7de1; padding:18px 22px; border-radius:0 12px 12px 0; box-shadow: 0 2px 8px rgba(42,125,225,0.08); margin-bottom:14px;">
  <p style="margin:0; color:#1a3b5c;">
    <b>③ La estructura de datos define el algoritmo.</b> En los algoritmos de búsqueda,
    cambiar únicamente el método de extracción
    (<code style="background:#eef4fc; color:#1a5ca8; padding:2px 6px; border-radius:4px;">pop(0)</code>
    vs <code style="background:#eef4fc; color:#1a5ca8; padding:2px 6px; border-radius:4px;">pop()</code>)
    transforma por completo el comportamiento de la búsqueda, sin alterar el resto del código.
  </p>
</div>

<div style="background:#f0f7ff; border-left:6px solid #2a7de1; padding:18px 22px; border-radius:0 12px 12px 0; box-shadow: 0 2px 8px rgba(42,125,225,0.08); margin-bottom:14px;">
  <p style="margin:0; color:#1a3b5c;">
    <b>④ El orden de los vecinos importa.</b> Tanto BFS como DFS dependen del orden en que
    aparecen los vecinos en el grafo. Reordenar la lista de adyacencia puede producir
    caminos distintos, especialmente en DFS.
  </p>
</div>

<div style="background:#f0f7ff; border-left:6px solid #2a7de1; padding:18px 22px; border-radius:0 12px 12px 0; box-shadow: 0 2px 8px rgba(42,125,225,0.08); margin-bottom:14px;">
  <p style="margin:0; color:#1a3b5c;">
    <b>⑤ Elegir el algoritmo depende del objetivo.</b> Si lo que se busca es el
    <b>camino más corto</b>, BFS es la elección correcta. Si se prioriza <b>ahorro de memoria</b>
    o simplemente <b>encontrar cualquier solución válida</b> en un grafo muy profundo,
    DFS puede ser preferible.
  </p>
</div>

<div style="background:#fff8ec; border-left:6px solid #e8a33d; padding:18px 22px; border-radius:0 12px 12px 0; box-shadow: 0 2px 8px rgba(232,163,61,0.10);">
  <p style="margin:0; color:#7a4a05;">
    <b>⑥ Limitaciones observadas.</b> Ninguno de los dos algoritmos usa información
    heurística del problema (distancias, tráfico, etc.). Ambos son <b>no informados</b>,
    por lo que en grafos grandes podrían volverse ineficientes. Una mejora natural sería
    incorporar costos por arista y aplicar algoritmos como <b>Dijkstra</b> o <b>A*</b>.
  </p>
</div>

<h2 style="color:#1e4a76; margin-top:28px;">5. Conclusión</h2>

<div style="background:#f0f7ff; border-left:6px solid #2a7de1; padding:18px 22px; border-radius:0 12px 12px 0; box-shadow: 0 2px 8px rgba(42,125,225,0.08);">
  <p style="margin:0 0 10px 0; color:#1a3b5c;">
    El laboratorio evidenció que los tres bloques no son independientes: la
    <b>representación vectorial</b> con NumPy prepara los datos, las <b>métricas</b>
    (producto escalar y distancia euclidiana) extraen información cuantitativa de ellos,
    y los <b>algoritmos de búsqueda</b> (BFS y DFS) permiten navegar estructuras relacionales
    derivadas de ese mismo tipo de datos. Juntos conforman el ciclo básico de cualquier
    agente inteligente: <b>percibir → representar → medir → decidir</b>.
  </p>
  <p style="margin:0; color:#1a3b5c;">
    La comparación BFS/DFS refuerza la idea de que <b>no existe un algoritmo universalmente
    mejor</b>, sino uno más adecuado según las restricciones del entorno: tamaño del grafo,
    memoria disponible, necesidad de optimalidad y presencia o ausencia de información
    adicional. Y en el caso de los datos, NumPy demuestra que <b>vectorizar no es solo
    cuestión de rendimiento, sino también de claridad conceptual</b>.
  </p>
</div>
