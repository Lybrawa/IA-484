<h2 style="color:#1e4a76; margin-top:24px;">Parte C: Implementación de DFS (Búsqueda en Profundidad)</h2>

<div style="background:#f0f7ff; border-left:6px solid #2a7de1; padding:16px 20px; border-radius:0 12px 12px 0; box-shadow: 0 2px 8px rgba(42,125,225,0.08); margin-bottom:20px;">
  <p style="margin:0; color:#1a3b5c;">
    DFS explora el grafo <b>en profundidad</b>: avanza por una rama hasta el final y luego
    <b>retrocede</b> (backtracking) para explorar las alternativas. Se implementa con una
    <b>pila LIFO</b> (<code style="background:#eef4fc; color:#1a5ca8; padding:2px 6px; border-radius:4px;">lista.pop()</code>)
    y un conjunto de <b>visitados</b>. <b>No garantiza el camino más corto</b>, pero requiere menos memoria que BFS en grafos profundos.
  </p>
</div>

<h2 style="color:#1e4a76; margin-top:28px;">Pseudocódigo</h2>

<div style="background:#0f2233; color:#d6e6ff; padding:18px 22px; border-radius:12px; font-family:'JetBrains Mono','Consolas',monospace; font-size:0.92em; line-height:1.7; box-shadow: 0 4px 12px rgba(15,34,51,0.25);">
  <span style="color:#7fb4ff;">DFS</span>(grafo, inicio, objetivo):<br>
  &nbsp;&nbsp;pila ← [[inicio]]<br>
  &nbsp;&nbsp;visitados ← ∅<br>
  &nbsp;&nbsp;<span style="color:#7fb4ff;">mientras</span> pila ≠ ∅:<br>
  &nbsp;&nbsp;&nbsp;&nbsp;camino ← pila.<span style="color:#9de0bd;">pop()</span><br>
  &nbsp;&nbsp;&nbsp;&nbsp;nodo_actual ← camino[-1]<br>
  &nbsp;&nbsp;&nbsp;&nbsp;<span style="color:#7fb4ff;">si</span> nodo_actual = objetivo: <span style="color:#7fb4ff;">retornar</span> camino<br>
  &nbsp;&nbsp;&nbsp;&nbsp;<span style="color:#7fb4ff;">si</span> nodo_actual ∉ visitados:<br>
  &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;visitados.<span style="color:#9de0bd;">add</span>(nodo_actual)<br>
  &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;<span style="color:#7fb4ff;">para</span> vecino en grafo[nodo_actual]:<br>
  &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;pila.<span style="color:#9de0bd;">append</span>(camino + [vecino])<br>
  &nbsp;&nbsp;<span style="color:#7fb4ff;">retornar</span> None
</div>
