<h2 style="color:#1e4a76; margin-top:24px;">Parte A: Representación de observaciones</h2>

<div style="background:#f0f7ff; border-left:6px solid #2a7de1; padding:16px 20px; border-radius:0 12px 12px 0; box-shadow: 0 2px 8px rgba(42,125,225,0.08); margin:20px 0;">
  <p style="margin:0; color:#1a3b5c;">
    Cada <b>observación</b> de un dataset simple (sensores de <b>temperatura</b> y
    <b>humedad</b>) se representa como un <b>vector</b> de dos componentes. Al agrupar las
    cuatro observaciones en una sola estructura, se obtiene una <b>matriz</b> de forma
    <code style="background:#eef4fc; color:#1a5ca8; padding:2px 6px; border-radius:4px;">(4, 2)</code>:
    4 filas (observaciones) × 2 columnas (variables).
  </p>
</div>
