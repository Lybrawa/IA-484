<h2 style="color:#1e4a76; margin-top:24px;">Parte A: Representación de observaciones</h2>

<div style="background:#f0f7ff; border-left:6px solid #2a7de1; padding:16px 20px; border-radius:0 12px 12px 0; box-shadow: 0 2px 8px rgba(42,125,225,0.08); margin:20px 0;">

<h2 style="color:#1e4a76; margin:0; font-weight:600;">
Código — Vectores y matriz del dataset
</h2>

</div>

```python
import numpy as np

obs1 = np.array([22.5, 60])   # temperatura, humedad
obs2 = np.array([19.0, 45])
obs3 = np.array([25.3, 70])
obs4 = np.array([21.0, 50])

dataset = np.array([obs1, obs2, obs3, obs4])

print(dataset)
print("Forma de la matriz (filas, columnas):", dataset.shape)
```

<h2 style="color:#1e4a76; margin-top:24px;">Parte B: Acceso a filas y columnas específicas</h2>
<div style="background:#f0f7ff; border-left:6px solid #2a7de1; padding:16px 20px; border-radius:0 12px 12px 0; box-shadow: 0 2px 8px rgba(42,125,225,0.08); margin:20px 0;">
  <p style="margin:0; color:#1a3b5c;">
    En NumPy se accede a elementos de una matriz usando la sintaxis
    <code style="background:#eef4fc; color:#1a5ca8; padding:2px 6px; border-radius:4px;">matriz[filas, columnas]</code>.
    Con el operador <code style="background:#eef4fc; color:#1a5ca8; padding:2px 6px; border-radius:4px;">:</code>
    se seleccionan <b>todos</b> los índices de esa dimensión (slicing).
  </p>
</div>
