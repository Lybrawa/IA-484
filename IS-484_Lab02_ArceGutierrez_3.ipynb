<h2 style="color:#1e4a76; margin-top:24px;">Parte A: Producto escalar entre dos observaciones</h2>

<div style="background:#f0f7ff; border-left:6px solid #2a7de1; padding:16px 20px; border-radius:0 12px 12px 0; box-shadow: 0 2px 8px rgba(42,125,225,0.08); margin:20px 0;">
  <p style="margin:0; color:#1a3b5c;">
    El <b>producto escalar</b> entre dos vectores <code style="background:#eef4fc; color:#1a5ca8; padding:2px 6px; border-radius:4px;">a</code> y
    <code style="background:#eef4fc; color:#1a5ca8; padding:2px 6px; border-radius:4px;">b</code> de la misma dimensión se define como la suma de los productos
    componente a componente:
    <code style="background:#eef4fc; color:#1a5ca8; padding:2px 6px; border-radius:4px;">a · b = Σ aᵢ · bᵢ</code>.
    En NumPy se calcula con <code style="background:#eef4fc; color:#1a5ca8; padding:2px 6px; border-radius:4px;">np.dot(a, b)</code>.
  </p>

Código — Producto escalar

```python
import numpy as np

obs1 = np.array([22.5, 60])   # temperatura, humedad
obs2 = np.array([19.0, 45])

producto = np.dot(obs1, obs2)
print("Producto escalar entre obs1 y obs2:", producto)
```

<h2 style="color:#1e4a76; margin-top:24px;">Parte B: Distancia euclidiana entre observaciones</h2>
<div style="background:#f0f7ff; border-left:6px solid #2a7de1; padding:16px 20px; border-radius:0 12px 12px 0; box-shadow: 0 2px 8px rgba(42,125,225,0.08); margin:20px 0;">
  <p style="margin:0; color:#1a3b5c;">
    La <b>distancia euclidiana</b> entre dos vectores <code style="background:#eef4fc; color:#1a5ca8; padding:2px 6px; border-radius:4px;">a</code> y
    <code style="background:#eef4fc; color:#1a5ca8; padding:2px 6px; border-radius:4px;">b</code> es la <b>norma L2</b> de su diferencia:
    <code style="background:#eef4fc; color:#1a5ca8; padding:2px 6px; border-radius:4px;">d(a, b) = √(Σ (aᵢ − bᵢ)²)</code>.
    En NumPy se calcula directamente con <code style="background:#eef4fc; color:#1a5ca8; padding:2px 6px; border-radius:4px;">np.linalg.norm(a − b)</code>.
  </p>
